```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df.to_csv("titanic.csv", index=False)
print(df.shape)
df.info()
```

```python
print(df.isnull().sum())
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df = df.drop(columns=["Cabin"])
```

```python
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
df["Survived"] = df["Survived"].astype("category")
df["Pclass"] = df["Pclass"].astype("category")
df["Sex"] = df["Sex"].astype("category")
```

```python
print(df["Age"].mean(), df["Age"].median())
print(df["Fare"].describe())
print(df["Sex"].value_counts())
print(df["Embarked"].value_counts())
```

```python
df.to_csv("titanic_cleaned.csv", index=False)
```

Task 1: Data Cleaning & Exploration with Pandas
For this task, I used the Titanic dataset, which has 891 rows and 12 columns. I loaded it into a pandas DataFrame and started by checking its shape and data types.

Next, I looked for missing values. The Age column had 177 missing entries, so I filled them with the median age, since the median is not affected by extreme values. Embarked had 2 missing entries, which I filled with the most common value (the mode). The Cabin column was missing most of its values (687 out of 891), so I dropped it because it would not give reliable information.

I then checked for duplicate rows and found none, so no rows were removed. After that, I converted Survived, Pclass and Sex to the category data type, as they represent groups rather than numbers.

Finally, I calculated basic statistics, including the mean and median of Age, a summary of Fare, and value counts for Sex and Embarked. This showed that the dataset has more male passengers than female, and that most passengers boarded at Southampton. I saved the cleaned data as titanic_cleaned.csv.
